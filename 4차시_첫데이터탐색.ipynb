## 5. 데이터 미리보기

## 6. 데이터 타입 확인

`info()`로 각 열의 데이터 타입과 결측치 개수를 확인합니다.

- `int64`: 정수
- `float64`: 실수
- `object`: 문자열
- `bool`: True/False

```python
# 각 열의 데이터 타입과 null 개수
df.info()
```

## 7. 기본 통계 보기

`describe()`로 숫자 열의 기본 통계를 확인합니다.

- `count`: 개수
- `mean`: 평균
- `std`: 표준편차
- `min/max`: 최소/최대
- `25%, 50%, 75%`: 사분위수

```python
# 숫자 열의 기본 통계
df.describe()
```

## 8. 값 분포 확인

`value_counts()`로 특정 열의 값 분포를 확인합니다.

```python
# 추천 여부 분포 확인
df['recommended'].value_counts()
```

```python
# 언어 분포 확인
df['language'].value_counts()
```

## 9. 결측치 확인

`isnull().sum()`으로 각 열별 결측치 개수를 확인합니다.

```python
# 결측치 개수 확인
df.isnull().sum()
```

```python
# 결측치 비율 확인
print("결측치 비율 (%)")
print((df.isnull().sum() / len(df) * 100).round(2))
```

## 10. 시각화 맛보기

matplotlib으로 추천/비추천 분포를 막대 그래프로 그려봅니다.

```python
# 한글 폰트 설정: Colab 기본 환경에서는 그래프의 한글이 □로 깨지므로 추가한 셀
!pip install -q koreanize-matplotlib
import koreanize_matplotlib
```

```python
import matplotlib.pyplot as plt

# 추천 여부 막대 그래프
df['recommended'].value_counts().plot(kind='bar', color=['#4CAF50', '#f44336'])
plt.title('추천/비추천 리뷰 분포')
plt.xlabel('추천 여부')
plt.ylabel('리뷰 수')
plt.xticks([0, 1], ['추천', '비추천'], rotation=0)
plt.show()
```

```python
# 플레이 시간 히스토그램
df['playtime_hours'].hist(bins=20, color='skyblue', edgecolor='black')
plt.title('플레이 시간 분포')
plt.xlabel('플레이 시간 (시간)')
plt.ylabel('리뷰 수')
plt.show()
```

```python
# 추가 탐색: 리뷰 길이, 중복, 이상한 값, 작성 기간
df['review_length'] = df['review_text'].str.len()
print(df['review_length'].describe().round(1))
print('중복된 리뷰 본문 수:', df['review_text'].duplicated().sum(), '/ 고유 본문 수:', df['review_text'].nunique())
print('플레이 시간 0 이하 리뷰 수:', (df['playtime_hours'] <= 0).sum())
print('작성 기간:', df['posted_date'].min(), '~', df['posted_date'].max())
print(df.groupby('recommended')['playtime_hours'].mean().round(1))
print(df['review_text'].value_counts().head(5))
```

```python
# 리뷰 길이 히스토그램
df['review_length'].hist(bins=20, color='orange', edgecolor='black')
plt.title('리뷰 길이 분포')
plt.xlabel('글자 수')
plt.ylabel('리뷰 수')
plt.show()
```

## 11. 탐색 결과 메모

**[데이터 개요]**
- 총 리뷰 수: 500개, 컬럼 7개 (숫자 3개, 텍스트 3개, True/False 1개)
- 언어: 전부 한국어(ko), 작성 기간: 2023-01-02 ~ 2025-12-30

**[발견한 특징]**
1. **추천 비율 63.4%**: 추천 317개, 비추천 183개로 추천이 약 1.7배 많다. 한쪽으로 심하게 쏠리지는 않아 두 그룹을 비교하기에 충분하다.
2. **결측치는 양호 수준**: playtime_hours 21개(4.2%), helpful_votes 15개(3.0%)가 비어 있다. 둘 다 0~5% 범위이고, 분석의 핵심인 review_text에는 결측치가 없다.
3. **같은 리뷰 본문이 매우 많이 반복된다**: review_id는 500개 모두 다르지만 고유한 본문은 82개뿐이다. 가장 많이 반복된 문장은 14번 등장한다. 본격적인 분석 전에 중복 처리 기준을 정해야 한다.
4. **추천한 사람이 훨씬 오래 플레이했다**: 평균 플레이 시간이 추천 45.9시간, 비추천 9.7시간이다. 플레이 시간이 0인 리뷰는 없다(최소 0.3시간).
5. **helpful_votes에 튀는 값이 있다**: 중앙값은 4인데 최댓값은 187이다. 소수 리뷰에 투표가 몰려 있어 평균(8.7)만 보면 오해할 수 있다.
6. **리뷰가 짧고 길이가 균일하다**: 16~37자, 평균 22.8자로 실제 Steam 리뷰보다 훨씬 짧다.
